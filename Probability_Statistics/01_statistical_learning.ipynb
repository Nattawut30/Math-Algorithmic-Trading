```python
""" Chapter 01: Statistical Learning """

# Y = f(X) + e.

# standard deviation = tells you how spread out the values are
# the low standard deviation = values cluster tightly around the average.
# the high standard deviation = they are spread wide. in production line, high standard deviation = inconsistent production

# f = some fixed but unknown function of X1, ... , Xp,
# e = a random error term which is independent of X and means 0
# f represents systematic information that X provides about Y.

# In general, the function f may involve more than one input variable.
# f is a 2D surface that must be estimated based on the observed data.
```

```python
""" Why Estimate f? """

# Y^ = f^(X)
# It is natural to seek to predict Y using X.

# The accuracy of Y^ as a prediction for Y depends on 2 quantities, which we will call the reducible error and the irreducible error.

# E(Y - Y^)^2 represents the average, or expected value, of the squared difference between the predicted and actual value of Y, and Var(e) represents the variance associated with the error term e.
# The irreducible error will always provide an upper bound on the accuracy of our prediction for Y. 
# This bound is almost always unknown in practice.

# Depending on whether our goal is prediction, inference o a combination of the 2 different methods for estimating f may be appropriate.
```

```python
""" How do we Estimate f?"""

# n = 30 data points. these obseved are called "the training data" we will use these observation to train or teach, our method how to estimate f
# Our end goal is to apply a statistical learning method to the training data in order to estimate the unknow function.
```

```python
""" Parametric Methods """

# We make an assumption about the function form, or shape of f.
# Instead of having to estimate an entirely arbitrary p-dimensional function f(x),
# one only needs to estimate the p+1 coefficients.

# After a model has been selected we need a procedure that uses the training data of fit or train the model
# It reduce the problem of estimating f down to one of estimating a set of parameters

# The potential disadvantage of a parametric approach is that the mdoel we choose will usually not match the true unknown form of f.
# If the choose model is too far from the true f, then our estimate will be poor.

# fitting a more flexible model requires estimating a greater number of parameters.
# These more complex models can lead to a phenomenon known as overfitting the data, which essentially means they follow the errors or noise too closely.

# The true f has some curvature that is not captured in the linear fit. 
```

```python
""" Non-Parametric Methods """

# They seek an estimate of f that gets as close to the data points as possible without being too rough or wiggly.
# Avoiding the assumption of a particular functional form of f, they have the potential to accurately fit a wider range of possible shapes for f.

# non-parametric methods disadvantage: since they do not reduce the problem if estimating f to a small number of parameters,
# a very large number of observations is needs in order to obtain an accurate estimate for f.

# To fit a thin-plate spline, I must select a level of smoothness.
```

```python
""" The Trade-Off Between Prediction Accuracy and Model Interpretability """

# Some method are less flexible or more restrictive in the sense that they can produce just a relatively small range of shapes to estimate f.
# If we are mainly intereted in inference, then restrictive models are much more interpretable.
# The new procedure is more restrictive in estimating the coefficients and sets a number of them to exactly 0
# The lasso is a less flexible aprroach than linear regression. It is also more interpretable than linear regression.
# The relationship between each predictor and the response is now modeled using a curve.

# Inference is a goal = using simple and relatively inflexible statistical learning methods.
```

```python
""" Supervised - Unsupervised Learning """

# I wish to fit a model that relates the response to the predictor, with the aim of accurately predicting the response for future observations or,
# better understanding the relationship between the response and the predictors.
# Many classical statistical learning methods such as linear regression and logistic regression,
# GAM, boosting and support vector machines, operate in the supervised learning domain.

# It is not possible to fit a linear regression model, since there is no response variable to predict.
# We are in some sense of working blindly just like the name referred to as "unsupervised" because we lack a response variable that can supervise our analysis.
# One statistical learning that we may use in this setting is "cluster analysis" or clustering.

# In practice, we often encounter data sets that contain many more than 2 variables.
# In this case, we cannot easily plot the observation. For instance, if there p variables in our data set,
# the p(p - 1) / 2 distinct scatterplots can be made and visual inspection is simply not a viable way to identify clusters.

# Automated clustering methods are important!
# The predictors can be measured relatively cheaply but the corresponding responses are much more expensive to collect.
# We refer to this setting as a "semi-supervised learning" problem.
```

```python
""" Regression - Classification Problems """

# As either quantitative or qualitative (also known as categorical)
# Quantitative vatiables take on numerical values = a person's age, height, or income, the value of a house, and the price of a stock.
# Qualitative variables take on values in one of K different classes or categories = a marital status, the brand of product purchased, debt or not, cancer diagnosis types.

# A quantitative = regression problems.
# A qualitative = classification problems.

# The predictors are qualitative or quantitative is generally considered less important!
```

```python
""" Assessing Model Accuracy """

# No one method dominates all others over all possible data sets.
# On a particular data set, one specific method may work best, but some method may work better on a similar but different data set.
# It is important to decide for any given set of data which method produces the besr results.
```

```python
""" Measuring the Quality of Fit """

# We need some way to measure how well its predictions actually match the observed data.
# We need to quantify the extent to which the predicted response value for a given observation is close to the true response value for that observation.

# Regression settings = MSE or mean squared error.

# The MSE will be small if the predicted responses are very close to the true responses,
# and will be large if for some of the observations, the predicted and true response differ substantially.

# We are interested in the accuracy of the predictions that we obtain when we apply our method to previousky unseen test data.
# A previously unseen test observation not used to train the statistical learning method.

# There is not guarantee that the method with the lowest training MSE will also have the lowest test MSE.
# The problem is that many statistical methods specifically estimate coefficients so as to minimize the training set MSE.
# For this reason, the training set MSE can be quite small but the test MSE is often much larger.

# As the level of flexibility increases, the curves fit the observed data more closely.
# Adjusting the level of flexibility of the smoothing spline fit, we can produce many different fits to this data.
# Linear Regression is at the most restrictive end, with 2 degress of freedom. The training MSE declines monotonically as flexibility increases.

# Overfitting the data. This happens because our statistical learning procedure is working too hard to find patterns in the training data, and may be picking up some patterns that are just...
# caused by random chance rather than by true properties of the unknown function f.
```

```python
# We almost always expect the training MSE to be smaller than the test MSE because most statistical learning methods either directly or indirectly seek to minimize the training MSE.
# Estimating the test MSE is considerably more difficult becuase mostly no test data are available!
```

```python
""" The Bias-Variance Trade-Off """

# The U-shape observed in the test MSE curves, turns out to be the result of two competing properties of statistical learning methods.
# We need to select a statistical learning method that simultaneously achieves low variance and low bias.
# Variance is inherently a non-negative quantity and squared bias is also non-negative. Expected test MSE can never lie below Var(e).

# Variance refers to the amount by which f^ would change if we estimated it using a different training data set, different training data sets will result in a different f^/
# We use more flexible methods, the variance will increase and the bias will decrease.
# At some point increasing flixibility has little impact on the bias but starts to significantly increase the variance.
```
