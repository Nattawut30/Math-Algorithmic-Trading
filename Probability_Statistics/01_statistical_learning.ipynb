```python
""" Why Estimate f? """

# Y^ = f^(X)
# It is natural to seek to predict Y using X.

# The accuracy of Y^ as a prediction for Y depends on 2 quantities, which we will call the reducible error and the irreducible error.

# E(Y - Y^)^2 represents the average, or expected value, of the squared difference between the predicted and actual value of Y, and Var(e) represents the variance associated with the error term e.
# The irreducible error will always provide an upper bound on the accuracy of our prediction for Y. 
# This bound is almost always unknown in practice.

# Depending on whether our goal is prediction, inference o a combination of the 2 different methods for estimating f may be appropriate.
```
